#### First technical decision.

We are gonna transform null -> 0.

**¿ What does it implies?**

That we are going to have a pre-processing engine that tranform those request with no Content-Length to zero.

```python
df= df.with_columns(
    pd.col("lenght")  # Using length column 
    .str.extract(r"(\d+)",1) # Extract digits using regular expressions
    .cast(pd.Int64, strict=False) # Transform those string digits to integers 64-bit
    .fill_null(0) # Filter where type is null, and convert it to zero.
)

# Showing results
df["lenght"].head()
```

```python
df["content"].head
```

```python
df= df.with_columns(
    pd.col("content")  # Using content column 
   .fill_null('') # Filter where type is null, and convert it to empty string.
)

# Showing results
df["content"].head()
```

```python
df= df.with_columns(
    pd.col("Accept")  # Using Accept column 
   .fill_null('') # Filter where Accept is null, and convert it to empty string.
)

# Showing results
df["Accept"].head()
```

```python
df["content-type"].head()
```

```python
df= df.with_columns(
    pd.col("content-type")  # Using Accept column 
   .fill_null('') # Filter where Accept is null, and convert it to empty string.
)

# Showing results
df["content-type"].head()
```
